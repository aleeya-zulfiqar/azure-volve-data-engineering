```python
from pyspark.sql import functions as F

base = "abfss://data@volvedatalake.dfs.core.windows.net"

bronze = spark.read.parquet(
    f"{base}/bronze/volve/parquet/production"
)

print("Rows:", bronze.count())
print("Columns:", len(bronze.columns))

bronze.printSchema()
bronze.show(5, truncate=False)
```

```python
from pyspark.sql import functions as F

silver = (
    bronze
    # Standardise column names
    .withColumnRenamed("DATEPRD", "prod_date")
    .withColumnRenamed("WELL_BORE_CODE", "well_bore_code")
    .withColumnRenamed("NPD_WELL_BORE_CODE", "well_bore_id")
    .withColumnRenamed("NPD_WELL_BORE_NAME", "well_name")
    .withColumnRenamed("NPD_FIELD_CODE", "field_id")
    .withColumnRenamed("NPD_FIELD_NAME", "field_name")
    .withColumnRenamed("NPD_FACILITY_CODE", "facility_id")
    .withColumnRenamed("NPD_FACILITY_NAME", "facility_name")
    .withColumnRenamed("ON_STREAM_HRS", "on_stream_hours")
    .withColumnRenamed("AVG_DOWNHOLE_PRESSURE", "avg_downhole_pressure")
    .withColumnRenamed("AVG_DOWNHOLE_TEMPERATURE", "avg_downhole_temperature")
    .withColumnRenamed("AVG_DP_TUBING", "avg_dp_tubing")
    .withColumnRenamed("AVG_ANNULUS_PRESS", "avg_annulus_pressure")
    .withColumnRenamed("AVG_CHOKE_SIZE_P", "avg_choke_size")
    .withColumnRenamed("AVG_CHOKE_UOM", "choke_uom")
    .withColumnRenamed("AVG_WHP_P", "avg_whp")
    .withColumnRenamed("AVG_WHT_P", "avg_wht")
    .withColumnRenamed("DP_CHOKE_SIZE", "dp_choke_size")
    .withColumnRenamed("BORE_OIL_VOL", "oil_volume")
    .withColumnRenamed("BORE_GAS_VOL", "gas_volume")
    .withColumnRenamed("BORE_WAT_VOL", "water_volume")
    .withColumnRenamed("BORE_WI_VOL", "water_injection_volume")
    .withColumnRenamed("FLOW_KIND", "flow_kind")
    .withColumnRenamed("WELL_TYPE", "well_type")

    # Date
    .withColumn("prod_date", F.to_date("prod_date"))

    # Clean text fields
    .withColumn("well_bore_code", F.trim("well_bore_code"))
    .withColumn("well_name", F.trim("well_name"))
    .withColumn("field_name", F.trim("field_name"))
    .withColumn("facility_name", F.trim("facility_name"))
    .withColumn("flow_kind", F.trim("flow_kind"))
    .withColumn("well_type", F.trim("well_type"))

    # Remove duplicate records
    .dropDuplicates()

    # Production records must have a date
    .filter(F.col("prod_date").isNotNull())
)

print("Bronze rows:", bronze.count())
print("Silver rows:", silver.count())
print("Rows removed:", bronze.count() - silver.count())

silver.printSchema()
silver.show(5, truncate=False)
```

```python
from pyspark.sql import functions as F

base = "abfss://data@volvedatalake.dfs.core.windows.net"

volve = spark.read.format("delta").load(
    f"{base}/silver/production"
)

print("Rows:", volve.count())
print("Columns:", len(volve.columns))

# Check whether well + date uniquely identifies a record
grain_check = (
    volve
    .groupBy("well_bore_id", "prod_date")
    .count()
    .filter(F.col("count") > 1)
)

print("Duplicate well/date combinations:", grain_check.count())

grain_check.show(10, truncate=False)
```

```python
volve.select(
    "well_bore_id",
    "well_name",
    "field_id",
    "field_name",
    "prod_date",
    "oil_volume",
    "gas_volume",
    "water_volume"
).orderBy("prod_date").show(10, truncate=False)

print("Date range:")
volve.select(
    F.min("prod_date").alias("min_date"),
    F.max("prod_date").alias("max_date")
).show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql import functions as F
from delta.tables import DeltaTable

gold_base = f"{base}/gold"

# -------------------------
# 1. dim_well
# -------------------------
dim_well = (
    volve
    .select(
        "well_bore_id",
        "well_bore_code",
        "well_name",
        "well_type",
        "facility_id",
        "facility_name"
    )
    .dropDuplicates(["well_bore_id"])
    .withColumn(
        "well_key",
        F.row_number().over(
            Window.orderBy("well_bore_id")
        )
    )
    .select(
        "well_key",
        "well_bore_id",
        "well_bore_code",
        "well_name",
        "well_type",
        "facility_id",
        "facility_name"
    )
)

# -------------------------
# 2. dim_field
# -------------------------
dim_field = (
    volve
    .select("field_id", "field_name")
    .dropDuplicates(["field_id"])
    .withColumn(
        "field_key",
        F.row_number().over(
            Window.orderBy("field_id")
        )
    )
    .select("field_key", "field_id", "field_name")
)

# -------------------------
# 3. dim_date
# -------------------------
date_range = volve.select(
    F.min("prod_date").alias("min_date"),
    F.max("prod_date").alias("max_date")
).first()

dim_date = (
    spark.range(
        1,
        (date_range["max_date"] - date_range["min_date"]).days + 2
    )
    .withColumn(
        "date",
        F.date_add(
            F.lit(date_range["min_date"]),
            F.col("id").cast("int") - 1
        )
    )
    .drop("id")
    .withColumn("date_key", F.date_format("date", "yyyyMMdd").cast("int"))
    .withColumn("year", F.year("date"))
    .withColumn("month", F.month("date"))
    .withColumn("month_name", F.date_format("date", "MMMM"))
    .withColumn("quarter", F.quarter("date"))
    .withColumn("day_of_month", F.dayofmonth("date"))
)

# -------------------------
# 4. dim_commodity
# -------------------------
dim_commodity = spark.createDataFrame(
    [
        (1, "OIL", "Oil"),
        (2, "GAS", "Gas"),
        (3, "WATER", "Water"),
        (4, "WATER_INJECTION", "Water Injection")
    ],
    ["commodity_key", "commodity_code", "commodity_name"]
)

print("dim_well:", dim_well.count())
print("dim_field:", dim_field.count())
print("dim_date:", dim_date.count())
print("dim_commodity:", dim_commodity.count())
```

```python
dim_well.write.format("delta").mode("overwrite").save(
    f"{gold_base}/dim_well"
)

dim_field.write.format("delta").mode("overwrite").save(
    f"{gold_base}/dim_field"
)

dim_date.write.format("delta").mode("overwrite").save(
    f"{gold_base}/dim_date"
)

dim_commodity.write.format("delta").mode("overwrite").save(
    f"{gold_base}/dim_commodity"
)

print("All Gold dimensions written successfully.")
```

```python
fact = (
    volve.alias("p")
    .join(
        dim_well.select("well_key", "well_bore_id").alias("w"),
        F.col("p.well_bore_id") == F.col("w.well_bore_id"),
        "left"
    )
    .join(
        dim_field.select("field_key", "field_id").alias("f"),
        F.col("p.field_id") == F.col("f.field_id"),
        "left"
    )
    .join(
        dim_date.select(
            "date_key", "date", "year"
        ).alias("d"),
        F.col("p.prod_date") == F.col("d.date"),
        "left"
    )
    .select(
        F.concat_ws(
            "_",
            F.col("p.well_bore_id"),
            F.date_format("p.prod_date", "yyyyMMdd")
        ).alias("production_key"),

        "w.well_key",
        "f.field_key",
        "d.date_key",

        F.col("p.prod_date").alias("production_date"),
        "d.year",

        F.col("p.on_stream_hours"),
        F.col("p.avg_downhole_pressure"),
        F.col("p.avg_downhole_temperature"),
        F.col("p.avg_dp_tubing"),
        F.col("p.avg_annulus_pressure"),
        F.col("p.avg_choke_size"),
        F.col("p.avg_whp"),
        F.col("p.avg_wht"),
        F.col("p.dp_choke_size"),

        F.col("p.oil_volume"),
        F.col("p.gas_volume"),
        F.col("p.water_volume"),
        F.col("p.water_injection_volume")
    )
)

print("Fact rows:", fact.count())
print("Fact columns:", len(fact.columns))
fact.show(5, truncate=False)
```

```python
from pyspark.sql.window import Window

# Window ordered by production date within each well
well_window = (
    Window
    .partitionBy("well_key")
    .orderBy("production_date")
)

fact = (
    fact
    .withColumn(
        "cumulative_oil_volume",
        F.sum("oil_volume").over(well_window)
    )
    .withColumn(
        "cumulative_gas_volume",
        F.sum("gas_volume").over(well_window)
    )
    .withColumn(
        "previous_oil_volume",
        F.lag("oil_volume").over(well_window)
    )
    .withColumn(
        "oil_decline_rate",
        F.when(
            (F.col("previous_oil_volume") > 0) &
            F.col("oil_volume").isNotNull(),
            1 - (
                F.col("oil_volume") /
                F.col("previous_oil_volume")
            )
        )
    )
    .drop("previous_oil_volume")
)

print("Fact rows:", fact.count())
fact.select(
    "production_key",
    "well_key",
    "production_date",
    "oil_volume",
    "cumulative_oil_volume",
    "oil_decline_rate"
).show(10, truncate=False)
```

```python
fact_path = f"{gold_base}/fact_production"

fact.write \
    .format("delta") \
    .mode("overwrite") \
    .partitionBy("year") \
    .save(fact_path)

print("fact_production written successfully.")
```

```python
fact_gold = spark.read.format("delta").load(fact_path)

print("Rows:", fact_gold.count())
print("Columns:", len(fact_gold.columns))
print("Years:", fact_gold.select("year").distinct().orderBy("year").collect())

fact_gold.show(5, truncate=False)
```

```python
updates = fact_gold.limit(10)

tgt = DeltaTable.forPath(
    spark,
    fact_path
)

(
    tgt.alias("t")
    .merge(
        updates.alias("s"),
        "t.production_key = s.production_key"
    )
    .whenMatchedUpdateAll()
    .whenNotMatchedInsertAll()
    .execute()
)

print("MERGE completed successfully.")
```
